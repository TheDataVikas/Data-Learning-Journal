```python
from pyspark.sql import functions as F
from pyspark.sql.types import *
```

```python
df = spark.read.table("de_practice.dbo.employees")
dept_df = spark.read.table("de_practice.dbo.departments")
```

```python
# Select just first_name and last_name from df.
(df.select("first_name", "last_name")).show()
```

```python
# Select all columns except manager_id.
df.drop("manager_id").show()
```

```python
# Select salary, but rename it to annual_salary in the output (use .alias()).
df.select(F.col("salary").alias("annual_salary")).show()
```
